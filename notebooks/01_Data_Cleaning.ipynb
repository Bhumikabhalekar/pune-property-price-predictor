```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/pune_property_raw.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
print(df[df["area"] == "area"].shape)
df = df[df["area"] != "area"]
```

```python
df = df.drop(columns=["Unnamed: 0"], errors="ignore")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print(df.shape)
```

```python
df = df.dropna(subset=["price", "area", "locality"])
print(df.shape)
```

```python
cols_to_drop = ["possesiondate", "roadfaceing", "overlooking",
                "projectname", "amenitiesnot", "pricepersquare"]
df = df.drop(columns=cols_to_drop)
print(df.shape)
```

```python
for c in ["area", "age", "floor", "totalfloor", "bhk", "price", "carpetarea", "additionalrooms"]:
    print(c, "->", df[c].dropna().unique()[:8])
```

```python
for col in ["area", "carpetarea"]:
    df[col] = (df[col].astype(str)
                      .str.replace(",", "", regex=False)
                      .str.extract(r"([\d.]+)")[0]
                      .astype(float))

df["price"] = pd.to_numeric(df["price"], errors="coerce")
```

```python
df["age"] = (df["age"].astype(str).str.findall(r"\d+")
               .str[-1].astype(float))

df["floor"] = df["floor"].astype(str).str.extract(r"(\d+)")[0].astype(float)
df["totalfloor"] = pd.to_numeric(df["totalfloor"], errors="coerce")
```

```python
df["bedrooms"] = df["bhk"].astype(str).str.extract(r"(\d+)")[0].astype(float)

df["property_type"] = (df["bhk"].astype(str)
                         .str.replace(r"^\s*\d+\s*(BHK|RK)\s*", "", regex=True)
                         .str.strip().str.lower())

df = df.drop(columns=["bhk"])
print(df["property_type"].value_counts())
```

```python
df["addi_rooms"] = (df["additionalrooms"].astype(str)
                      .str.extract(r"(\d+)")[0].astype(float).fillna(0))
df = df.drop(columns=["additionalrooms"])
```

```python
for col in ["facing", "locality", "neworold", "ownership", "status"]:
    df[col] = df[col].astype(str).str.strip().str.lower().replace("nan", np.nan)
```

```python
df.info()
df.describe().T
```

```python
for c in ["balconies", "bathroom", "opensides"]:
    print(c, "->", df[c].dropna().unique()[:15])
```

```python
for c in ["balconies", "bathroom", "opensides"]:
    df[c] = df[c].astype(str).str.extract(r"(\d+)")[0].astype(float)
```

```python
keep = ["apartment", "villa", "independent house", "independent floor", "penthouse"]
df = df[df["property_type"].isin(keep)].reset_index(drop=True)
print(df.shape)
```

```python
df["price_per_sqft"] = df["price"] / df["area"]
df["price_per_sqft"].describe()
```

```python
print(df.nsmallest(5, "price_per_sqft")[["locality", "area", "price", "price_per_sqft"]])
print(df.nlargest(5, "price_per_sqft")[["locality", "area", "price", "price_per_sqft"]])
```

```python
for c in ["balconies", "bathroom", "opensides"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df[["balconies", "bathroom", "opensides"]].dtypes
```

```python
df["price_per_sqft"].quantile([0.01, 0.05, 0.5, 0.95, 0.99])
```

```python
low, high = df["price_per_sqft"].quantile([0.01, 0.99])
print("keeping price per sqft between", round(low), "and", round(high))

before = len(df)
df = df[df["price_per_sqft"].between(low, high)].reset_index(drop=True)
print("removed", before - len(df), "rows | shape now:", df.shape)
```

```python
df["area"].quantile([0.001, 0.01, 0.5, 0.99, 0.999])
```

```python
amen = (df["amenitiesavailable"].dropna()
          .str.split(",").explode().str.strip())
amen = amen[amen != ""]
amen_counts = amen.value_counts()
print(len(amen_counts), "different amenities")
amen_counts
```

```python
import re

amen_sets = df["amenitiesavailable"].fillna("").apply(
    lambda s: {x.strip() for x in s.split(",")})

# keep amenities that appear in at least 2% of properties
common = amen_counts[amen_counts >= 0.02 * len(df)].index

am_cols = []
for a in common:
    name = "am_" + re.sub(r"[^a-z0-9]+", "_", a.lower()).strip("_")
    df[name] = amen_sets.apply(lambda s: int(a in s))
    am_cols.append(name)

df["amenity_count"] = df[am_cols].sum(axis=1)
df = df.drop(columns=["amenitiesavailable"])
print(len(am_cols), "amenity columns created")
df[am_cols + ["amenity_count"]].head()
```

```python
print(df.shape)
df["amenity_count"].describe()
```

```python
df = df.drop(columns=["carpetarea"])
```

```python
for c in ["facing", "ownership", "status"]:
    df[c] = df[c].fillna("unknown")
```

```python
print("floor higher than building height:", (df["floor"] > df["totalfloor"]).sum())
print("bedrooms = 0:", (df["bedrooms"] == 0).sum())
print("bathrooms > bedrooms + 3:", (df["bathroom"] > df["bedrooms"] + 3).sum())
```

```python
df = df.drop(columns=["price_per_sqft"])   # helper only, the model must not see it
df.to_csv("../data/processed/pune_property_clean.csv", index=False)
print(df.shape)
```
